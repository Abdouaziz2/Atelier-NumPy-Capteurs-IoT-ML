# Partie 5 — Filtrage booléen

Le filtrage booléen permet de sélectionner uniquement les valeurs d'un tableau qui respectent une condition.

On utilise une condition qui produit un tableau de True et False.


**1) Afficher les températures supérieures à 30 °C**


```python
# Crée une condition booléenne :
# True pour les températures > 30, False sinon
condition = temperatures_journee > 30

# Utilise cette condition pour filtrer le tableau
temperatures_sup_30 = temperatures_journee[condition]

temperatures_sup_30
```

2) Afficher les températures « normales » càd comprises entre 20 °C et 30 °C


```python
# Vérifie les deux conditions :
# température >= 20 ET température <= 30
condition_normale = (
    (temperatures_journee >= 20)
    & (temperatures_journee <= 30)
)

# On Sélectionne uniquement les températures normales
temperatures_normales = temperatures_journee[condition_normale]

temperatures_normales
```

**Opérateurs utilisés**

>= : supérieur ou égal.

<= : inférieur ou égal.

& : ET logique élément par élément.

⚠️ Avec NumPy, pour combiner plusieurs conditions, on utilise & et on met chaque condition entre parenthèses.

L'écart-type mesure également la dispersion des données.


```python
temperature_ecart_type = np.std(temperatures_capteurs)

temperature_ecart_type
```

La différence importante :

**2. Standardisation**

La standardisation transforme les données afin qu'elles aient généralement :

moyenne ≈ 0
écart-type ≈ 1

La formule est :

z = (x - moyenne) / écart-type

En NumPy :

```python
moyenne = np.mean(temperatures_capteurs)
ecart_type = np.std(temperatures_capteurs)

temperatures_standardisees = (
    (temperatures_capteurs - moyenne)
    / ecart_type
)

temperatures_standardisees
```

Verifications

**Verifications**

```python
np.mean(temperatures_standardisees)
```

```python
np.std(temperatures_standardisees)
```

**Pourquoi standardiser ?**

En Machine Learning, certaines variables peuvent avoir des échelles très différentes.

**Par exemple :**

Température       → 20 à 35
Pression          → 950 à 1050
Consommation      → 0 à 10000

Sans transformation, certaines variables peuvent avoir une influence disproportionnée dans certains algorithmes.

La standardisation permet de mettre les variables sur une échelle comparable.

35. Normalisation Min-Max

La normalisation Min-Max transforme les valeurs généralement dans l'intervalle :

[0, 1]

Formule :

(x - minimum) / (maximum - minimum)

En NumPy :

```python
temperature_min = np.min(temperatures_capteurs)
temperature_max = np.max(temperatures_capteurs)

temperatures_normalisees = (
    (temperatures_capteurs - temperature_min)
    / (temperature_max - temperature_min)
)

temperatures_normalisees
```

Verifications

```python
np.min(temperatures_normalisees)
```

```python
np.max(temperatures_normalisees)
```

**Standardisation vs normalisation**

Standardisation :

moyenne ≈ 0
écart-type ≈ 1

Min-Max :

minimum = 0
maximum = 1

Ce sont deux techniques différentes de mise à l'échelle des données.

**Partie 9 — Broadcasting**

Le broadcasting est une fonctionnalité de NumPy qui permet d'effectuer une opération entre des tableaux de dimensions différentes, lorsque leurs dimensions sont compatibles.

Ici, on va appliquer une correction différente à chacun des 3 capteurs, sans écrire de boucle.

**1. Créer temperatures_3x3**

On veut représenter :

3 capteurs
3 heures : 10h, 11h et 12h

Donc notre tableau aura la forme :

```python
# Générateur aléatoire reproductible
rng = np.random.default_rng(42)

# Températures de 3 capteurs entre 20 °C et 35 °C
temperatures_3x3 = rng.uniform(
    20,
    35,
    size=(3, 3)
)

temperatures_3x3
```

2. Créer les corrections

L'énoncé nous donne :

```python
correction = np.array([1, -0.5, 0.2])
```

**3. Appliquer le Broadcasting**

```python
mesures_corrigees = temperatures_3x3 + correction

mesures_corrigees
```

**Pourquoi parle-t-on de Broadcasting ?**

Parce que NumPy étend conceptuellement le petit tableau correction pour pouvoir effectuer l'opération avec le tableau 3 × 3.

Il n'est pas nécessaire de créer manuellement :
